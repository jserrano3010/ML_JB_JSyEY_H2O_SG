# Dataset de un sitio (plan B)

Este entregable se ejecutó con la serie real de NASA POWER de la celda de Sabanagrande (1981–2025), descargada en `data/raw/Entregable1_Sabanagrande.xlsx`. Este notebook la convierte al formato del proyecto y calcula ET0. El diseño regional (notebook 00) queda para el segundo entregable.

**Limitación declarada:** 16.436 observaciones, por debajo de las 20.000 sugeridas por la guía.

```python
import sys, warnings
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from src import config as cfg
np.random.seed(cfg.SEED)
plt.rcParams["figure.dpi"] = 110
from src.hidrologia import et0_penman_monteith
df = pd.read_excel(cfg.DATA_RAW / "Entregable1_Sabanagrande.xlsx", sheet_name="Datos NASA")
df = df.replace(-999, np.nan)
df["punto"], df["tipo_punto"] = cfg.PUNTO_OBJETIVO, "finca_prueba"
df["lat"], df["lon"], df["elevacion_m"] = cfg.LAT_FINCA, cfg.LON_FINCA, np.nan
df["ET0"] = et0_penman_monteith(df.T2M_MAX, df.T2M_MIN, df.RH2M, df.WS2M, df.ALLSKY_SFC_SW_DWN, lat=cfg.LAT_FINCA,
                                dia_juliano=df.fecha.dt.dayofyear, elevacion_m=0.0, presion_kpa=df.PS).round(3)
cols = ["fecha", "punto", "tipo_punto", "lat", "lon", "elevacion_m", *cfg.POWER_PARAMS, "ET0"]
df = df[cols].sort_values("fecha").reset_index(drop=True)
assert df.fecha.min() == pd.Timestamp("1981-01-01") and df.fecha.max() == pd.Timestamp("2025-12-31")
assert df.duplicated(["punto", "fecha"]).sum() == 0
df.to_csv(cfg.DATA_PROCESSED / cfg.ARCHIVO_DATOS, index=False)
print(df.shape); df.isna().mean().mul(100).round(2)
```
